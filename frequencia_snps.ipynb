```python
import pandas as pd
import os

COLUNAS_REMOCAO = ["FID", "PAT", "MAT", "SEX", "PHENOTYPE"]
caminho = ('Banco Genetico Imputado LD')
df_completo = None
for arquivo in os.listdir(caminho):
    if arquivo.endswith('.raw'):
        df_gen = pd.read_csv(f'{caminho}/{arquivo}',sep=r'\s+')
        df_gen = df_gen.drop(columns=COLUNAS_REMOCAO)
        if df_completo is None:
            df_completo = df_gen
        else:
            df_completo = pd.merge(left=df_completo,right=df_gen,on='IID',how='inner')
```

```python
frequencias_valores = df_completo.isin([1, 2]).sum().to_dict()
frequencias_0 = (df_completo ==0).sum().to_dict()
```

```python
frequencias_porcentagem = (df_completo.isin([1, 2]).sum() / len(df_completo)).round(2).to_dict()
```

```python
print('Quantidade de snps menores que a frequencia')
print(f'{'Frequencia':<15}|{'Quantidade':>15}')
início = 0.1
fim = 0.51
passo = 0.02

i = início
while i <= fim:
    quantidade = sum(1 for snp in df_completo.columns if frequencias_porcentagem[snp] < i)
    print(f'{i:<15.2f}|{quantidade:>15}')
    i += passo
```
